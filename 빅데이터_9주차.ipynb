```python
x = np.arange(-20,21)
y1 = 2*x
y2 = x**2 + 5
y3 = -.2*x**2 - 5

plt.plot(x,y1,'r--', x,y2,'g^-', x, y3, 'b*:')
plt.axis([-30,30,-30,30])
plt.show()
```

```python
x = np.arange(-20,21)
y1 = 2*x
y2 = x**2 + 5
y3 = -.2*x**2 - 5

plt.plot(x,y1,'r--', label='2*x')
plt.plot(x,y2,'g^-', label='x**2 + 5')
plt.plot(x, y3, 'b*:', label='-.2*x**2 - 5')
plt.legend()

plt.axis([-30,30,-30,30])
plt.show()
```

```python
n = 30
normal_d = np.random.randn(1000)
rnadom_d = np.random.randn(1000) * 6 - 3

plt.hist(normal_d, n, color='orange', label='normal')

plt.hist(rnadom_d, n, color='blue',alpha=0.3, label='random dist')
plt.legend()
```

```python
mu1, sigma1 = 10,2
mu2, sigma2 = -6,3


srandard_gauss = np.random.randn(1000)
gauss1 = mu1 + sigma1 * np.random.randn(1000)
gauss2 = mu2 + sigma2 * np.random.randn(1000)

plt.figure(figsize=(10,6))
plt.hist(srandard_gauss, bins=50,alpha=0.4)
plt.hist(gauss1, bins=50,alpha=0.4)
plt.hist(gauss2, bins=50,alpha=0.4)
plt.show()
```
