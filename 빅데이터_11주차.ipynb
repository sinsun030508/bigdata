```python
np.random.seed(42)
data = np.random.randn(2,100)

fig, ax = plt.subplots(2,2, figsize=(5,4))

ax[0,0].hist(data[0])
ax[1,0].scatter(data[0],data[1])
ax[0,1].plot(data[0],data[1])
ax[1,1].hist2d(data[0], data[1])
plt.show()
```

```python
import seaborn as sns
sns.set_theme(style="darkgrid")
tips = sns.load_dataset("tips")
tips.head()
tips.shape
sns.histplot(tips['tip'], kde=False, bins=10)
```

```python
sns.relplot(data=tips, x="total_bill", y="tip", col='time', hue='smoker', style='smoker')
plt.show()
```

```python
tips['tup_pct']=100*tips['tip']/tips['total_bill']
sns.histplot(tips['tip'], kde=False, bins=20)
```

```python
sns.relplot(x='total_bill', y = 'tip', hue='smoker', style='smoker', data = tips)
```

```python
ax = sns.regplot(data = tips, x='size', y = 'total_bill')
ax.set_xlabel('Size')
ax.set_ylabel('Total Bill')
ax.set_title('Total Bill and Size')
plt.show()
```
