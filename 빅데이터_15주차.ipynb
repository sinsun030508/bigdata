```python
import urllib.request
import json

# =========================================================
# [필수 수정] 여기에 네이버에서 발급받은 본인의 키를 넣으세요!
client_id = "5YrXMTr0o1jmJDpTzzhz"
client_secret = "ZQH734MPrA"
# =========================================================

def get_naver_news_api(keyword):
    encText = urllib.parse.quote(keyword)
    # display=5 로 상위 5개 지정
    url = f"https://openapi.naver.com/v1/search/news.json?query={encText}&display=5&sort=sim"

    request = urllib.request.Request(url)
    request.add_header("X-Naver-Client-Id", client_id)
    request.add_header("X-Naver-Client-Secret", client_secret)

    try:
        response = urllib.request.urlopen(request)
        rescode = response.getcode()

        if rescode == 200:
            response_body = response.read()
            data = json.loads(response_body.decode('utf-8'))

            print(f"=== '{keyword}' 관련 네이버 뉴스 상위 5개 (API 방식) ===\n")

            for idx, item in enumerate(data['items'], 1):
                # 제목에 포함된 불필요한 HTML 태그 정리
                title = item['title'].replace("", "").replace("", "").replace("&quot;", '"').replace("&amp;", "&")
                link = item['link']

                print(f"{idx}. {title}")
                print(f"   링크: {link}")
                print("-" * 60)
        else:
            print(f"오류 발생: 상태 코드 {rescode}")

    except urllib.error.HTTPError as e:
        if e.code == 401:
            print("에러 발생: HTTP Error 401 (Unauthorized)")
            print("-> Client ID 또는 Client Secret 값이 네이버 개발자 센터의 값과 일치하는지 다시 확인해 주세요. 공백이 포함되었을 수도 있습니다.")
        else:
            print(f"HTTP 에러 발생: {e}")
    except Exception as e:
        print(f"기타 에러 발생: {e}")

# 실행 테스트
search_keyword = "인공지능"
get_naver_news_api(search_keyword)
```
