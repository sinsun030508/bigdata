```python
import seaborn as sns

anscombe = sns.load_dataset('anscombe').head(200)
anscombe.query("dataset == 'I'")
```

```python
#Ci=None을 표시하지 않고 진짜값이 있을 가능성인 높은 값
sns.lmplot(x='x', y='y', data=anscombe.query("dataset == 'I'"),
           ci=None, scatter_kws={'s': 80})
```
